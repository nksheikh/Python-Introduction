#### "Assertion" String Methods

There are several string methods that allow logical tests related to what's captured in the string. For instance:

```python
print(mystring.isalpha()) # Tests whether string is made of letters
print(mystring.isdigit()) # Tests whether string is made up of numbers
print(mystring.islower()) # Tests whether all letters are lowercase
print(mystring.isupper()) # Tests whether all letters are uppercase
```

Note that the "isupper()" and "islower() methods ignore non-alpha characters, though they do require at least one:

```python
print("test123".islower())
print("123".isupper())
print("T123".isupper())
```

#### "Re-format" String Methods

There are numerous other methods designed to adjust or reformat strings:

```python
print(mystring.upper())
print(mystring.capitalize())
print(mystring.encode('utf8'))
```

The first two methods are likely self-explanatory. The third illustrates how to move to binary from a decoded string. The "b" in front of 'analytics' indicates this is a binary (or "bytes" representation). If you encounter errors that complain about decoding errors or "bytes-like" objects, they likely indicate your string variable is not formatted correctly. The rule to remember is that you <u>encode</u> **to bytes** and <u>decode</u> **from bytes**.

It is important to point out that these methods do not alter the original variable. This would require re-assignment:

```python
print(mystring)
mystring = mystring.encode('utf8')
print(mystring)
mystring = mystring.decode('utf8')
print(mystring)
```

#### More "Re-format" methods

There are several methods that I use very frequently when working with string variables:
- <b>split</b>: using a delimiter, splits the string into elements which are returned as a list (the *list* method <b>join</b> is the inverse). The default delimiter is whitespace.
- <b>replace</b>: replaces a given character string with another
- <b>strip</b>: typically used to remove whitespace, though another character can be specified
    
Let's illustrate:

```python
myname = "Robbie Moon"
print(myname.split())
print(myname.replace('Moon','Dufus'))
print(myname.strip())
```

##### <i> Custom Functions </i>
Python makes it incredibly easy to write custom functions. Before we do so, it's important to recognize that *indendation* is vital to working Python code. This will make more sense once we define a custom function.

Task: Write a function that formats floating point decimals into feet and inches
- accepts one argument, a float
- returns a dictionary with two values, "feet" and "inches"

```python
# To define a function, start with a "def" line like so:
def format_height(height):
    # Everything within the function must be indented
    # define feet and inches like we did above:
    feet = int(height)
    inches = round((height-feet)*12)
    # function finishes with "return" (can have more than one return)
    return {'feet':feet,'inches':inches}

print(format_height(height))
```

Note that the function could have returned that data in a number of ways. Rather than using a dictionary with two elements, feet and inches, I could have returned two variables, or a tuple, or list, etc. 

## Logic Control
Logic control refers to the commands we use to control how our script proceeds, depending on various conditions we establish. It is extremely important for processing any data, and particularly when we attempt to structure unstructured data. We'll start with the most basic form of logic control, if/elif/else (aka "if/then/else").

### if/elif/else
"If" statements are the most basic form of logic control. They operate similar to the "IF" function in excel. Key parts:
- if ____: The first condition to be checked
- elif ____: Equivalent to "else if"; must be preceded by "if" statement
- else: what to do if previous conditions fail

<b>Important</b>:
- To test for equality, <u>must use a double "="</u> (or ==). Other options:
>- can also use greater than (>), less than (<), greater/less than or equal to (>= or <=)
>- modulo (%)
- <u>Indentation determines code blocks</u> (example will illustrate)

Let's look at a basic if/elif/else block:

```python
new_student = "Jose"
# let's check to see if "Jose" is in our students_test dictionary
if new_student in student_test.keys():
    print("Jose is here")
else:
    print("Jose is not here")
```

Blocks like this are helpful with dictionaries because attempting to access an element not present produces an error:

```python
print(student_test['Jose'])
```

Incorporating "elif" simply adds to the logic chain. The only requirement is that an logic block like this start with "if" (else is actually optional)

```python
print(students)
print(student_test)
if new_student in student_test.keys():
    print("Jose is here")
elif new_student in students:
    print("Jose is in students but not student_test")
else:
    print("Jose is missing!")
```

The conditions you use for these types of control blocks are very flexible. Here's another example:

```python
odd = 7
even = 4
if odd % 2 == 0:
    print(f"{odd} is even")
else:
    print(f"{odd} is odd")
```

For loops are most common method for iterating over objects, such as files, integers, websites, words, etc. They take the format <b>for <i>i</i> in <i>iterable</i></b>, where:
- <i>i</i>: a <u>temporary</u> variable that references each element of the <i>iterable</i>
- <i>iterable</i>: the "list" of things we are iterating over; could be an explicit list (like a list of files) or a function of method that <i>generates</i> a list ("range()" is very common)

Let's set up an example that illustrates both of these. 

Referring to the data set up earlier, suppose I want to prepare a loop that iterates over the students in my class (list), checks for a grade, and if there prints the grade. Otherwise, it let's me know the test hasn't been taken.

My grader gives me the following data, which I load in a dictionary (the grader doesn't see student names):

```python
anon_scores = {1002:95,1003:75,1004:83}
```

I can now set up the loop I describe above (pay special attention to indentation):

```python
for student in students: # "i" is student, and "students" is my iterable
    # check to see if student has taken exam
    exam_key = student_test[student]
    if exam_key in anon_scores.keys(): # note that the "keys() method returns a set of dictionary keys"
        grade = anon_scores[exam_key]
        print(f"{student} received a {grade} on the exam.")
    else:
        print(f"{student} has not taken the exam yet.")
        
```

In other cases, we may wish to iterate some task 100 or 1000 times, maybe as part of a larger simulation. Let's set up a very basic simulation that we run 1000 times, where we flip a coin. We can accomplish this with a loop of this format:

```python
import numpy as np

coin_flips = [] # we'll call a heads 1, and tails 0
for i in range(1000):
    rand = np.random.random()
    if rand >= 0.50:
        coin_flips.append(1)
    else:
        coin_flips.append(0)

print(f"Heads was flipped {np.mean(coin_flips)*100: .2f}% of times")
```

There are a few other important commands that are useful in loops:
- pass: Move on in script but do nothing at this point
- continue: Move to next iteration of loop
- break: discontinue loop immediately

Suppose we've decided that if our coin produces the same result 5 times in a row, it's not fair. The odds of this happening in a 3 flip sequence is only 3.125% (0.5^5), but it may happen in this 1000 flip sequence. Let's see how we'd structure our loop to test this:

```python
consecutives = 0
heads = 1
for i in range(1000):
    if np.random.random()>=0.5:
        heads = 1
    else:
        heads = 0
        
    if i==0:
        last_flip = heads
        pass
    
    else:
        if last_flip == heads:
            consecutives+=1
        
        else:
            consecutives = 0
            last_flip = heads
    
    if consecutives == 5:
        print(f"Coin is bad!!! Made it to iteration {i}")
        break
    else:
        continue # This isn't necessary, but we'll include to illustrate all commands
```

The "gamblers fallacy" refers to the fact that individuals have a tendency to ignore the independence of random events, assuming that a given sequence should mirror the original expected value for a set.
